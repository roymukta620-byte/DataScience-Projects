iloc means:

integer location

It is used to access rows/columns by POSITION NUMBER.

Syntax:

df.iloc[row_index]

or

df.iloc[row_index, column_index]

So in our code:

df.commute_time.iloc[4]

means:

go to column commute_time
take the value at position/index 4
(the 5th row, because counting starts from 0)

From our dataframe:

index	commute_time
0	50
1	15
2	27
3	10
4	20min

So:

df.commute_time.iloc[4]

returns:

"20min"

Then:

type(df.commute_time.iloc[4])

checks its datatype.

we can also do:

df.iloc[4]

→ gives the ENTIRE 5th row.

Or:

df.iloc[4, 2]

→ row 4, column 2.

Very heavily used in:

data cleaning
ML preprocessing
debugging
feature engineering
inspecting specific rows.

```python
def update_commute_times(val):
  splits = ['min','hr','hour']
  for substr in splits:
    if substr in val:
      split_idx = val.find(substr)
      new_val = float(val[:split_idx].strip())
      return new_val if substr == 'min' else 60.0*new_val

  substr = 'apx.'
  if substr in val:
    split_idx = val.find(substr)
    return float(val[split_idx + len(substr):].strip())

  if 'campus' in val:
    return 0

  return float(val)

tests = ['25', '20mins', '1hr', '2 hr', '3 hours', '4hours', '5-campus', '6 on-campus', 'apx.8','apx. 9']
print(tests)
print([update_commute_times(test) for test in tests])
```

```python
df.commute_mins = df.commute_mins.apply(update_commute_times)
```

```python
def update_height(val):
  substr = 'cm'
  if substr in val:
    split_idx = val.find(substr)
    return float(val[:split_idx].strip())

  substr = '.'
  if substr in val:
    if val.find(substr) == 1:
      return float(val) * 100.0
    return float(val)

  substr = ','
  if substr in val:
    val = val.replace(substr,'.')
    return float(val) * 100.0

  substr = "\'"
  if substr in val:
    split_strs = val.replace("'"," ").replace('"','').split()
    feet = float(split_strs[0])
    inches = float(split_strs[1])

    return 2.54 * inches + 30.48*feet

  return float(val)
```
